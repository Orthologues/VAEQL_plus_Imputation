## Ampute all these datasets using the package <code>pyampute</code>

```python
complete_awan_datasets: dict[str, pd.DataFrame] = {
    "breast_cancer": breast_cancer_df,
    "travel": travel_df,
    "spamhouse": spamhouse_df,
    "parkinsons": parkinsons_df,
    "letter_recognition": letter_df,
    "news_popularity": news_df,
    "credit_card_df": credit_card_df
}
```

All the feature values in these datasets are numerical, thus there is not need for further pre-processing

```python
def create_amputed_MAR_or_MNAR_dataset_and_export(
    output_root_folder: str,
    output_ds_name: str,
    index: int, 
    missing_rate: float, 
    input_df: pd.DataFrame,
    missing_mechanism: str
):

    amputed_df = input_df.copy()

    for col in input_df.columns:
        if missing_mechanism not in {"MAR", "MNAR"}:
            raise Exception("Sorry, the missing mechanism must be either 'MAR' or 'MNAR'")

        missing_pattern = \
        [{
            "incomplete_vars": [col],
            "mechanism": missing_mechanism
        }]
        amputor = MultivariateAmputation(prop=missing_rate, patterns=missing_pattern)
        amputed_df[col] = amputor.fit_transform(input_df)[col].values

    # output the CSV file
    amputed_df.to_csv(f'{output_root_folder}/{output_ds_name}/{missing_mechanism}_{int(missing_rate*100)}_perc_{index}.csv', index=False)
```

```python
missing_rate = [0.05*rate for rate in range(1, 6)]
```

```python
OUTPUT_ROOT = "/work/jiz_imputation/Awan_2022_amputed_datasets"
```

```python
# --- wrapper for parallel jobs ---
def amputation_worker(args):
    output_root_folder, ds_name, index, missing_rate, df, mech = args
    create_amputed_MAR_or_MNAR_dataset_and_export(output_root_folder, ds_name, index, missing_rate, df, mech)
    return f"{ds_name} | {mech} | rate={missing_rate:.2f} | index={index} done"
```

```python
# --- build job list ---
output_root = "/work/jiz_imputation/Awan_2022_amputed_datasets"

missing_rates = [0.05 * rate for rate in range(1, 6)]
indices = range(11)
mechanisms = ["MAR", "MNAR"]

job_params_list = []

for ds_name, df in complete_awan_datasets.items():
    if not path_exists(f"{output_root}/{ds_name}"):
        mkdir(f"{output_root}/{ds_name}")
    for mech in mechanisms:
        for rate in missing_rates:
            for idx in indices:
                job_params_list.append((output_root, ds_name, idx, rate, df, mech))

# --- run multiprocessing ---
num_cores = min(60, cpu_count()-2)  # use up to 60 or however many are available minus 2
with Pool(processes=num_cores) as pool:
    for result in pool.imap_unordered(amputation_worker, job_params_list):
        print(result)
```
